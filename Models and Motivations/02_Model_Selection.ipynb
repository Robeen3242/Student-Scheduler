# ✅ = Recommended | ❌ = Not Recommended

## ✅ Logistic Regression
Our base-line model. We measure the performance of our other models based off of this one.

Expectiations: Good on small data, maybe worse with large data. Simple models can outperform complex ones, we may see that here.

## ✅ Decision Trees / Random Forest
A very simple nonlinear model which we can then show the improvement on in Random Forest.

Expectations: Since our data is small, it is unclear whether Random Forest will provide any meaningful improvements (if any) over Decision Trees. In this case, we will choose one or the other (but still test both) and see whether reducing overfitting is worth added model complexity.

## ✅ XGBoost
An ensemble model that builds upon Decision Trees by correcting mistakes made by previous trees. Rather than relying on many independent trees like Random Forest, XGBoost focuses on improving difficult predictions throughout training.

Expectations: Since our engineered features may contain complex relationships between workload, sleep, stress, and upcoming deadlines, I expect XGBoost to be a strong candidate. However, the increased model complexity may also make it more susceptible to overfitting if not tuned carefully.

## ❌ Neural Networks
We just simply do not have enough data to support a Neural Network. Backprop would not be very beneficial with our limted data.

Expectations: If we did implement, overfitting and generalization is a huge concern unless we obtain more data.

## ❌ SVM
For our problem we want the model to adapt to changing student behaviour over time. Burnout is influenced by many factors beyond academics, and the relationship between our engineered features and burnout may evolve throughout a semester or even between semesters. Although SVMs can model nonlinear decision boundaries through Radial or Polynomial kernels, they are fundamentally designed to learn a fixed decision boundary from the training data. Our problem instead emphasizes personalization and continual adaptation, making SVM a less natural fit than some of the other approaches.

Expectations: With the best kernel settings I don't expect SVM to actually perform badly on classification or finding good hyperplane. In fact, it may have a very good fitting hyperplane to our data. Burnout evolves throughout the semester though as it is a feeling a student gets, so in a sense it is a constantly shifting hyperplane. Although it these shifts might not be large, does it necessarily perform better than regression?

## ❌ KNN
Even now I can see the project growing in feature complexity, if we were to add more features and/or more data, distance might not be a useful metric to classify.

Expectations: KNN might work for our data set right now, but as our data set grows larger I expect KNN to degrade in performance rapidly.

## ❌ Naive Bayes
Features from our daily rating are correlated/not independent. Having exams or assignments can influence these daily ratings and so Naive Bayes would look at Sleep, Stress, and Energy as independent sources of burnout which isn't exactly the context of our problem. Even if we were to make a "burnout score" that calculates the probability of burnout based on the daily rating features we still have data such as `exams_within` and `tasks_within` that are not as easy to consolidate into one value.

Expectations: The model is just not suited for our problem and is expected to underperform compared to models that learn feature interactions. While it is possible to model our data to work with Naive Bayes is it really worth the time and effort?
