## Preprocessing

Even before getting into data cleaning and actual training we have some important issues to address:

- Standard data splitting (randomize data -> split into test/valid/train sets) does not work because the nature of our problem requires our data to be sequenced in order by day
- Each semester does not contain a lot of data

We have two solutions and we can compare them. Also I'm not using mathematical notation to stroke mathemtician dick here:

### Heap of Semesters

Batch 1: Sem 1

Batch 2: Sem 1 + Sem 2

Batch 3: Sem 1 + Sem 2 + Sem 3

Batch n: Sem 1 + Sem 2 + Sem 3 + ... + Sem n

### Windowed Stacks

Batch 1: Sem 1 + Sem 2

Batch 2: Sem 2 + Sem 3

Batch 3: Sem 3 + Sem 4

Batch n: Sem n + Sem n+1

## Additional Preprocessing Problems

In my feature design and my implementation of it in the frontend, I kept NaaN values in mind and I beleive I have done a good job in mitigating missing values under standard conditions (user sometimes fills / doesn't fill daily ratings.) 

One oversight we did make though was what it means when features like `max_priority` that depend on the existence of exams or tasks in order to function.

Another oversight is what the program does once there is no more exams left for `days_until_next_exam`. A simple change by making -1 indicate this is a perfectly reasonable fix but this is not to say it won't affect our models. Linear regression may take the -1 literally as a numeric value while trees may be able to isolate it easily.
