## Preprocessing

Even before getting into data cleaning and actual training we have some important issues to address:

- Standard data splitting (randomize data -> split into test/valid/train sets) does not work because the nature of our problem requires our data to be sequenced in order by day
- Each semester does not contain a lot of data

We have two solutions and we can compare them. Also I'm not using mathematical notation to stroke mathemtician dick here:

### Heap of Semesters

Batch 1: Sem 1

Batch 2: Sem 1 + Sem 2

Batch 3: Sem 1 + Sem 2 + Sem 3

Batch n: Sem 1 + Sem 2 + Sem 3 + ... + Sem n

### Windowed Stacks

Batch 1: Sem 1 + Sem 2

Batch 2: Sem 2 + Sem 3

Batch 3: Sem 3 + Sem 4

Batch n: Sem n + Sem n+1
